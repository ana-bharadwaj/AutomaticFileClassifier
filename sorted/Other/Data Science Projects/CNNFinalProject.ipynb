````python
import os, glob, warnings
import numpy as np
import pandas as pd
import librosa
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, BatchNormalization,
    Dropout, Dense, GlobalAveragePooling2D
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping

import kagglehub

warnings.filterwarnings("ignore")
np.random.seed(42)
tf.random.set_seed(42)

dataset_path = kagglehub.dataset_download("imsparsh/musicnet-dataset")
metadata = pd.read_csv(os.path.join(dataset_path, "musicnet_metadata.csv"))

def load_labels(path):
    dfs = []
    for f in glob.glob(os.path.join(path, "*.csv")):
        df = pd.read_csv(f)
        df["id"] = int(os.path.basename(f).replace(".csv", ""))
        dfs.append(df)
    return pd.concat(dfs, ignore_index=True)

train_labels = load_labels(os.path.join(dataset_path, "musicnet/musicnet/train_labels"))
test_labels  = load_labels(os.path.join(dataset_path, "musicnet/musicnet/test_labels"))

data = pd.concat([
    metadata.merge(train_labels, on="id"),
    metadata.merge(test_labels, on="id")
])

composer_dict = {
    None: 0, "Bach": 1, "Beethoven": 2, "Brahms": 3,
    "Cambini": 4, "Dvorak": 5, "Faure": 6,
    "Haydn": 7, "Mozart": 8, "Ravel": 9, "Schubert": 10
}

def extract_tempo(m):
    m = str(m).lower()
    for t in ["presto", "allegro", "andante", "adagio", "largo"]:
        if t in m:
            return t
    return None

tempo_dict = {None: 0, "presto": 1, "allegro": 2, "andante": 3, "adagio": 4, "largo": 5}

data["composer"] = data["composer"].map(composer_dict)
data["tempo"] = data["movement"].apply(extract_tempo).map(tempo_dict)

instrument_map = (
    data.groupby("id")["instrument"]
    .agg(lambda x: x.value_counts().index[0])
    .to_dict()
)

data = data.drop_duplicates("id")
data["instrument"] = data["id"].map(instrument_map)

id2instr = dict(zip(data["id"].astype(str), data["instrument"]))
id2comp  = dict(zip(data["id"].astype(str), data["composer"]))
id2tempo = dict(zip(data["id"].astype(str), data["tempo"]))

def preprocess_audio(path, seg_len=3.0, n_seg=4):
    y, sr = librosa.load(path, sr=None)
    y = np.append(y[0], y[1:] - 0.97 * y[:-1])

    seg_samples = int(seg_len * sr)
    if len(y) < seg_samples:
        segments = [y]
    else:
        starts = np.linspace(0, len(y) - seg_samples, n_seg).astype(int)
        segments = [y[s:s+seg_samples] for s in starts]

    mels = []
    for seg in segments:
        S = librosa.feature.melspectrogram(
            y=seg, sr=sr, n_mels=64, hop_length=1024
        )
        mels.append(librosa.power_to_db(S))

    return mels

X, yi, yc, yt = [], [], [], []

for split in ["train_data", "test_data"]:
    audio_dir = os.path.join(dataset_path, f"musicnet/musicnet/{split}")
    for f in os.listdir(audio_dir):
        if not f.endswith(".wav"):
            continue
        fid = f.replace(".wav", "")
        if fid not in id2instr:
            continue
        for mel in preprocess_audio(os.path.join(audio_dir, f)):
            X.append(mel)
            yi.append(id2instr[fid])
            yc.append(id2comp[fid])
            yt.append(id2tempo[fid])

max_T = max(m.shape[1] for m in X)
n_mels = X[0].shape[0]

X_pad = np.zeros((len(X), n_mels, max_T))
for i, x in enumerate(X):
    X_pad[i, :, :x.shape[1]] = x

X_pad = X_pad[..., np.newaxis]

yi = to_categorical(yi)
yc = to_categorical(yc)
yt = to_categorical(yt)

X_train, X_test, yi_tr, yi_te, yc_tr, yc_te, yt_tr, yt_te = train_test_split(
    X_pad, yi, yc, yt, test_size=0.2, random_state=42
)

def create_model(input_shape, n_classes):
    inp = Input(shape=input_shape)

    x = Conv2D(32, (3,3), activation="relu", padding="same")(inp)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.2)(x)

    x = Conv2D(64, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.2)(x)

    x = Conv2D(128, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.3)(x)

    x = Conv2D(256, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = Dropout(0.3)(x)

    x = GlobalAveragePooling2D()(x)
    x = Dropout(0.5)(x)

    out = Dense(n_classes, activation="softmax")(x)
    return Model(inp, out)

def plot_training_curves(history, title):
    plt.figure(figsize=(12,4))

    plt.subplot(1,2,1)
    plt.plot(history.history["accuracy"], label="Train")
    plt.plot(history.history["val_accuracy"], label="Val")
    plt.title(f"{title} Accuracy")
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(history.history["loss"], label="Train")
    plt.plot(history.history["val_loss"], label="Val")
    plt.title(f"{title} Loss")
    plt.legend()

    plt.show()

def train_eval(Xtr, Xte, ytr, yte, name):
    model = create_model(Xtr.shape[1:], ytr.shape[1])
    model.compile(
        optimizer=Adam(1e-3),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    callbacks = [
        ReduceLROnPlateau(patience=3, factor=0.5, verbose=1),
        EarlyStopping(patience=6, restore_best_weights=True)
    ]

    history = model.fit(
        Xtr, ytr,
        validation_data=(Xte, yte),
        epochs=40,
        batch_size=32,
        callbacks=callbacks,
        verbose=1
    )

    plot_training_curves(history, name)

    y_pred = np.argmax(model.predict(Xte), axis=1)
    y_true = np.argmax(yte, axis=1)

    print(f"\n{name} Accuracy:", accuracy_score(y_true, y_pred))
    print(classification_report(y_true, y_pred, digits=4))

    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"{name} Confusion Matrix")
    plt.show()

train_eval(X_train, X_test, yi_tr, yi_te, "Instrument")
train_eval(X_train, X_test, yc_tr, yc_te, "Composer")
train_eval(X_train, X_test, yt_tr, yt_te, "Tempo")
````

# Define the Convolutional Neural Network (CNN) model architecture.
def create_model(input_shape, n_classes):
    """
    Creates a CNN model for audio classification.

    Args:
        input_shape (tuple): The shape of the input data (e.g., (n_mels, max_T, 1)).
        n_classes (int): The number of output classes for the classification task.

    Returns:
        tf.keras.models.Model: A compiled Keras Model.
    """
    # Input layer defines the shape of the incoming data.
    inp = Input(shape=input_shape)

    # First Convolutional Block
    # Conv2D: 32 filters, 3x3 kernel, ReLU activation, 'same' padding to preserve dimensions.
    x = Conv2D(32, (3,3), activation="relu", padding="same")(inp)
    x = BatchNormalization()(x) # Normalizes activations of the previous layer.
    x = MaxPooling2D((2,2))(x)  # Reduces spatial dimensions, downsampling the feature maps.
    x = Dropout(0.2)(x)         # Randomly sets 20% of input units to 0 to prevent overfitting.

    # Second Convolutional Block
    x = Conv2D(64, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.2)(x)

    # Third Convolutional Block
    x = Conv2D(128, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.3)(x)         # Increased dropout rate.

    # Fourth Convolutional Block
    x = Conv2D(256, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = Dropout(0.3)(x)

    # Global Average Pooling Layer
    # Reduces each feature map to a single number by taking the average,
    # effectively flattening the output while retaining spatial information.
    x = GlobalAveragePooling2D()(x)
    x = Dropout(0.5)(x)         # Higher dropout before the dense output layer.

    # Output Layer
    # Dense layer with 'softmax' activation for multi-class classification,
    # outputting probabilities for each class.
    out = Dense(n_classes, activation="softmax")(x)
    # Construct the model with specified input and output.
    return Model(inp, out)

def train_eval(Xtr, Xte, ytr, yte, name):
    """
    Trains a CNN model, evaluates its performance, and displays results.

    Args:
        Xtr (np.ndarray): Training features.
        Xte (np.ndarray): Test features.
        ytr (np.ndarray): Training labels (one-hot encoded).
        yte (np.ndarray): Test labels (one-hot encoded).
        name (str): A descriptive name for the classification task (e.g., "Instrument", "Composer").
    """
    # Create a new model instance for the specific classification task.
    model = create_model(Xtr.shape[1:], ytr.shape[1])
    # Compile the model with the Adam optimizer, categorical cross-entropy loss, and accuracy metric.
    model.compile(
        optimizer=Adam(1e-3), # Learning rate of 0.001
        loss="categorical_crossentropy", # Suitable for multi-class classification with one-hot encoded labels.
        metrics=["accuracy"] # Monitor accuracy during training.
    )

    # Define callbacks to adjust learning rate and stop training early.
    callbacks = [
        # Reduce learning rate when validation loss stops improving.
        ReduceLROnPlateau(patience=3, factor=0.5, verbose=1),
        # Stop training if validation loss does not improve for a certain number of epochs,
        # and restore the best weights observed during training.
        EarlyStopping(patience=6, restore_best_weights=True)
    ]

    # Train the model.
    history = model.fit(
        Xtr, ytr,
        validation_data=(Xte, yte), # Provide validation data for monitoring.
        epochs=40, # Maximum number of epochs.
        batch_size=32, # Number of samples per gradient update.
        callbacks=callbacks, # Apply the defined callbacks.
        verbose=1 # Show training progress.
    )

    # Plot the training and validation accuracy and loss curves.
    plot_training_curves(history, name)

    # Make predictions on the test set.
    y_pred = np.argmax(model.predict(Xte), axis=1) # Get the class with the highest probability.
    y_true = np.argmax(yte, axis=1) # Convert one-hot encoded true labels back to integer labels.

    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues") # Use a heatmap for better visualization.

````python
import os, glob, warnings
import numpy as np
import pandas as pd
import librosa
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, BatchNormalization,
    Dropout, Dense, GlobalAveragePooling2D
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping

import kagglehub

warnings.filterwarnings("ignore")
np.random.seed(42)
tf.random.set_seed(42)

dataset_path = kagglehub.dataset_download("imsparsh/musicnet-dataset")
metadata = pd.read_csv(os.path.join(dataset_path, "musicnet_metadata.csv"))

def load_labels(path):
    dfs = []
    for f in glob.glob(os.path.join(path, "*.csv")):
        df = pd.read_csv(f)
        df["id"] = int(os.path.basename(f).replace(".csv", ""))
        dfs.append(df)
    return pd.concat(dfs, ignore_index=True)

train_labels = load_labels(os.path.join(dataset_path, "musicnet/musicnet/train_labels"))
test_labels  = load_labels(os.path.join(dataset_path, "musicnet/musicnet/test_labels"))

data = pd.concat([
    metadata.merge(train_labels, on="id"),
    metadata.merge(test_labels, on="id")
])

composer_dict = {
    None: 0, "Bach": 1, "Beethoven": 2, "Brahms": 3,
    "Cambini": 4, "Dvorak": 5, "Faure": 6,
    "Haydn": 7, "Mozart": 8, "Ravel": 9, "Schubert": 10
}

def extract_tempo(m):
    m = str(m).lower()
    for t in ["presto", "allegro", "andante", "adagio", "largo"]:
        if t in m:
            return t
    return None

tempo_dict = {None: 0, "presto": 1, "allegro": 2, "andante": 3, "adagio": 4, "largo": 5}

data["composer"] = data["composer"].map(composer_dict)
data["tempo"] = data["movement"].apply(extract_tempo).map(tempo_dict)

instrument_map = (
    data.groupby("id")["instrument"]
    .agg(lambda x: x.value_counts().index[0])
    .to_dict()
)

data = data.drop_duplicates("id")
data["instrument"] = data["id"].map(instrument_map)

id2instr = dict(zip(data["id"].astype(str), data["instrument"]))
id2comp  = dict(zip(data["id"].astype(str), data["composer"]))
id2tempo = dict(zip(data["id"].astype(str), data["tempo"]))

def preprocess_audio(path, seg_len=3.0, n_seg=4):
    y, sr = librosa.load(path, sr=None)
    y = np.append(y[0], y[1:] - 0.97 * y[:-1])

    seg_samples = int(seg_len * sr)
    if len(y) < seg_samples:
        segments = [y]
    else:
        starts = np.linspace(0, len(y) - seg_samples, n_seg).astype(int)
        segments = [y[s:s+seg_samples] for s in starts]

    mels = []
    for seg in segments:
        S = librosa.feature.melspectrogram(
            y=seg, sr=sr, n_mels=64, hop_length=1024
        )
        mels.append(librosa.power_to_db(S))

    return mels

X, yi, yc, yt = [], [], [], []

for split in ["train_data", "test_data"]:
    audio_dir = os.path.join(dataset_path, f"musicnet/musicnet/{split}")
    for f in os.listdir(audio_dir):
        if not f.endswith(".wav"):
            continue
        fid = f.replace(".wav", "")
        if fid not in id2instr:
            continue
        for mel in preprocess_audio(os.path.join(audio_dir, f)):
            X.append(mel)
            yi.append(id2instr[fid])
            yc.append(id2comp[fid])
            yt.append(id2tempo[fid])

max_T = max(m.shape[1] for m in X)
n_mels = X[0].shape[0]

X_pad = np.zeros((len(X), n_mels, max_T))
for i, x in enumerate(X):
    X_pad[i, :, :x.shape[1]] = x

X_pad = X_pad[..., np.newaxis]

yi = to_categorical(yi)
yc = to_categorical(yc)
yt = to_categorical(yt)

X_train, X_test, yi_tr, yi_te, yc_tr, yc_te, yt_tr, yt_te = train_test_split(
    X_pad, yi, yc, yt, test_size=0.2, random_state=42
)

def create_model(input_shape, n_classes):
    inp = Input(shape=input_shape)

    x = Conv2D(32, (3,3), activation="relu", padding="same")(inp)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.2)(x)

    x = Conv2D(64, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.2)(x)

    x = Conv2D(128, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2,2))(x)
    x = Dropout(0.3)(x)

    x = Conv2D(256, (3,3), activation="relu", padding="same")(x)
    x = BatchNormalization()(x)
    x = Dropout(0.3)(x)

    x = GlobalAveragePooling2D()(x)
    x = Dropout(0.5)(x)

    out = Dense(n_classes, activation="softmax")(x)
    return Model(inp, out)

def plot_training_curves(history, title):
    plt.figure(figsize=(12,4))

    plt.subplot(1,2,1)
    plt.plot(history.history["accuracy"], label="Train")
    plt.plot(history.history["val_accuracy"], label="Val")
    plt.title(f"{title} Accuracy")
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(history.history["loss"], label="Train")
    plt.plot(history.history["val_loss"], label="Val")
    plt.title(f"{title} Loss")
    plt.legend()

    plt.show()

def train_eval(Xtr, Xte, ytr, yte, name):
    model = create_model(Xtr.shape[1:], ytr.shape[1])
    model.compile(
        optimizer=Adam(1e-3),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    callbacks = [
        ReduceLROnPlateau(patience=3, factor=0.5, verbose=1),
        EarlyStopping(patience=6, restore_best_weights=True)
    ]

    history = model.fit(
        Xtr, ytr,
        validation_data=(Xte, yte),
        epochs=40,
        batch_size=32,
        callbacks=callbacks,
        verbose=1
    )

    plot_training_curves(history, name)

    y_pred = np.argmax(model.predict(Xte), axis=1)
    y_true = np.argmax(yte, axis=1)

    print(f"\n{name} Accuracy:", accuracy_score(y_true, y_pred))
    print(classification_report(y_true, y_pred, digits=4))

    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"{name} Confusion Matrix")
    plt.show()

train_eval(X_train, X_test, yi_tr, yi_te, "Instrument")
train_eval(X_train, X_test, yc_tr, yc_te, "Composer")
train_eval(X_train, X_test, yt_tr, yt_te, "Tempo")
````
